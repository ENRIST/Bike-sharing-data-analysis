## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

```python
def cek_outlier_iqr(df, cols):
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        outliers = df[(df[col] < lower) | (df[col] > upper)]
        print(f"Outlier {col}: {len(outliers)} baris (batas wajar: {lower:.4f} s.d. {upper:.4f})")
```

```python
cek_outlier_iqr(day_df, ['temp', 'atemp', 'hum', 'windspeed', 'casual', 'registered', 'cnt'])
```

**Steps to Take:**
1. **Kesalahan Tipe Data (`dteday`):** Kolom `dteday` pada `day_df` dan `hour_df` masih bertipe `object` (string), bukan `datetime`.
   - **Solusi:** Mengonversi tipe data kolom `dteday` menjadi `datetime64` menggunakan `pd.to_datetime()`.
2. **Inaccurate Value (`hum == 0`):** Ditemukan nilai kelembapan udara (`hum`) sebesar `0.0` pada tanggal `2011-03-10` (1 baris di `day_df` dan 22 baris di `hour_df`) meskipun kondisi cuaca tercatat berkabut (`weathersit = 2`) hingga hujan/salju ringan (`weathersit = 3`). Secara fisik kelembapan `0%` saat kondisi hujan tidak masuk akal dan merupakan kesalahan pencatatan sensor.
   - **Solusi:** Mengubah nilai `0.0` pada kolom `hum` menjadi `NaN`, lalu mengisinya menggunakan metode interpolasi linier (`.interpolate()`) agar urutan waktu tetap terjaga tanpa menghapus baris.
3. **Nilai Kode & Normalisasi Kurang Deskriptif:** Kolom kategorikal (`season`, `yr`, `workingday`, `weathersit`) masih berupa kode angka (`0–1` atau `1–4`), serta kolom `temp` masih bernilai ternormalisasi (dibagi nilai maksimum `41`°C).
   - **Solusi:** Menambahkan kolom label teks deskriptif dan kolom suhu aktual dalam Celsius (`temp * 41`) pada tahap *cleaning* agar hasil analisis dan visualisasi mudah dipahami.

**Insight:** (Opsional)
- Tidak ditemukan *missing value* (`NaN`), data duplikat, maupun ketidakkonsistenan penjumlahan (`casual + registered == cnt`) pada `day_df` maupun `hour_df`.
- Ditemukan **2 permasalahan kualitas data utama**, yaitu kesalahan tipe data pada kolom `dteday` (`object` pada kedua tabel) dan *inaccurate value* berupa `hum == 0.0` di tanggal `2011-03-10` (1 baris pada `day_df` dan 22 baris pada `hour_df`).
- Pada `hour_df` juga tercatat 2.180 baris dengan `windspeed == 0.0` (kemungkinan kecepatan angin terlalu lemah untuk terbaca anemometer) serta *outlier* statistik pada kolom `casual`, `registered`, dan `cnt` (misal 44 baris `casual` di `day_df` dan 505 baris `cnt` di `hour_df`). Nilai *outlier* penyewaan tersebut tetap dipertahankan karena merupakan lonjakan permintaan nyata pada akhir pekan dan jam sibuk, bukan kesalahan input.

### Cleaning Data

#### Fixing day_df problem

```python
day_clean = day_df.copy()
```

```python
day_clean['dteday'] = pd.to_datetime(day_clean['dteday'])
```

```python
day_clean['hum'] = day_clean['hum'].replace(0, np.nan).interpolate()
```

```python
season_map = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weather_map = {
    1: 'Clear/Partly Cloudy',
    2: 'Mist/Cloudy',
    3: 'Light Snow/Rain',
    4: 'Heavy Rain/Ice/Fog'
}

day_clean['season_label'] = day_clean['season'].map(season_map)
day_clean['weather_label'] = day_clean['weathersit'].map(weather_map)
day_clean['year_label'] = day_clean['yr'].map({0: 2011, 1: 2012})
day_clean['day_type'] = day_clean['workingday'].map({1: 'Working Day', 0: 'Weekend/Holiday'})
day_clean['temp_c'] = day_clean['temp'] * 41
```

```python
print("Tipe dteday day_clean:", day_clean['dteday'].dtype)
print("Sisa hum == 0:", (day_clean['hum'] == 0).sum())
print("Nilai hum pada 2011-03-10:", day_clean.loc[day_clean['dteday'] == '2011-03-10', 'hum'].values)
print("Total missing value:", day_clean.isna().sum().sum())
```

#### Fixing hour_df problem

```python
hour_clean = hour_df.copy()
```

```python
hour_clean['dteday'] = pd.to_datetime(hour_clean['dteday'])
```

```python
hour_clean['hum'] = hour_clean['hum'].replace(0, np.nan).interpolate()
```

```python
hour_clean['season_label'] = hour_clean['season'].map(season_map)
hour_clean['weather_label'] = hour_clean['weathersit'].map(weather_map)
hour_clean['year_label'] = hour_clean['yr'].map({0: 2011, 1: 2012})
hour_clean['day_type'] = hour_clean['workingday'].map({1: 'Working Day', 0: 'Weekend/Holiday'})
hour_clean['temp_c'] = hour_clean['temp'] * 41
```

```python
print("Tipe dteday hour_clean:", hour_clean['dteday'].dtype)
print("Sisa hum == 0:", (hour_clean['hum'] == 0).sum())
print("Total missing value:", hour_clean.isna().sum().sum())
hour_clean[['dteday', 'hr', 'season_label', 'weather_label', 'day_type', 'temp_c', 'hum', 'cnt']].head(3)
```

**Insight:** (Opsional)

## Visualization & Explanatory Analysis

### Pertanyaan 1:
Bagaimana perbandingan rata-rata dan persentase perubahan jumlah penyewaan sepeda harian (`cnt`) berdasarkan empat musim (`season`) serta kondisi cuaca (`weathersit`) sepanjang periode Januari 2011–Desember 2012 untuk menentukan jadwal pemeliharaan armada dan strategi promosi musiman?

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

# Grafik 1A: Rata-rata Penyewaan Harian per Musim
season_avg = day_clean.groupby('season_label')['cnt'].mean().sort_values(ascending=False).reset_index()
colors_season = ['#1f77b4' if i == 0 else ('#d62728' if i == len(season_avg)-1 else '#d3d3d3') for i in range(len(season_avg))]

bars0 = axes[0].bar(season_avg['season_label'], season_avg['cnt'], color=colors_season, width=0.6)
axes[0].set_title('Rata-rata Penyewaan Harian Tertinggi di Fall, Terendah di Spring', fontsize=12, fontweight='bold', pad=12)
axes[0].set_xlabel('Musim', fontsize=11)
axes[0].set_ylabel('Rata-rata Penyewaan Harian (Sepeda)', fontsize=11)
axes[0].set_ylim(0, 6500)
for bar in bars0:
    h = bar.get_height()
    axes[0].annotate(f"{h:,.0f}", (bar.get_x() + bar.get_width()/2, h + 120), ha='center', fontsize=10, fontweight='bold')
axes[0].spines[['top', 'right']].set_visible(False)

# Grafik 1B: Rata-rata Penyewaan Harian per Kondisi Cuaca
weather_avg = day_clean.groupby('weather_label')['cnt'].mean().sort_values(ascending=False).reset_index()
colors_weather = ['#1f77b4' if i == 0 else ('#d62728' if i == len(weather_avg)-1 else '#d3d3d3') for i in range(len(weather_avg))]

bars1 = axes[1].bar(weather_avg['weather_label'], weather_avg['cnt'], color=colors_weather, width=0.55)
axes[1].set_title('Penyewaan Anjlok 63% Saat Hujan/Salju Ringan Dibanding Cuaca Cerah', fontsize=12, fontweight='bold', pad=12)
axes[1].set_xlabel('Kondisi Cuaca', fontsize=11)
axes[1].set_ylabel('Rata-rata Penyewaan Harian (Sepeda)', fontsize=11)
axes[1].set_ylim(0, 5800)
for bar in bars1:
    h = bar.get_height()
    axes[1].annotate(f"{h:,.0f}", (bar.get_x() + bar.get_width()/2, h + 120), ha='center', fontsize=10, fontweight='bold')
axes[1].spines[['top', 'right']].set_visible(False)

plt.tight_layout()
plt.show()
```

## Analisis Lanjutan (Opsional)

### Clustering Non-Machine Learning (Manual Grouping & Binning)

**Tujuan Analisis:**
Untuk menerjemahkan temuan pola jam dan pengaruh suhu menjadi segmen operasional yang siap dieksekusi tanpa algoritma *machine learning*, dilakukan pengelompokan (*clustering*) menggunakan dua teknik berikut:
1. **Manual Grouping pada Jam (`time_cluster`):** Membagi 24 jam operasional ke dalam 4 klaster waktu (*Morning Rush 06–09*, *Midday 10–15*, *Evening Rush 16–19*, dan *Night/Early Morning 20–05*) guna menentukan waktu prioritas *rebalancing* armada.
2. **Binning pada Suhu Aktual (`temp_bin`):** Mengelompokkan suhu harian (`temp_c`) ke dalam 3 kategori (*Cool <15°C*, *Mild 15–25°C*, dan *Warm >25°C*) untuk mengukur sensitivitas permintaan terhadap suhu lingkungan.

```python
# Manual Grouping untuk Klaster Waktu (hour_clean)
def categorize_hour(hr):
    if 6 <= hr <= 9:
        return 'Morning Rush (06-09)'
    elif 10 <= hr <= 15:
        return 'Midday (10-15)'
    elif 16 <= hr <= 19:
        return 'Evening Rush (16-19)'
    else:
        return 'Night/Early Morning (20-05)'

hour_clean['time_cluster'] = hour_clean['hr'].apply(categorize_hour)
```

```python
# Binning untuk Kategori Suhu Harian (day_clean)
day_clean['temp_bin'] = pd.cut(
    day_clean['temp_c'],
    bins=[0, 15, 25, 45],
    labels=['Cool (<15°C)', 'Mild (15-25°C)', 'Warm (>25°C)']
)
```

```python
# Visualisasi Hasil Analisis Lanjutan
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))

cluster_order = ['Morning Rush (06-09)', 'Midday (10-15)', 'Evening Rush (16-19)', 'Night/Early Morning (20-05)']
sns.barplot(
    data=hour_clean, x='time_cluster', y='cnt', hue='day_type',
    order=cluster_order, estimator='mean', errorbar=None,
    palette={'Working Day': '#1f77b4', 'Weekend/Holiday': '#ff7f0e'}, ax=axes[0]
)
axes[0].set_title('Evening Rush Paling Padat di Hari Kerja; Midday di Akhir Pekan', fontsize=12, fontweight='bold', pad=12)
axes[0].set_xlabel('Klaster Waktu Operasional', fontsize=11)
axes[0].set_ylabel('Rata-rata Penyewaan per Jam', fontsize=11)
axes[0].set_ylim(0, 480)
axes[0].tick_params(axis='x', rotation=12)
for p in axes[0].patches:
    if p.get_height() > 0:
        axes[0].annotate(f"{p.get_height():.0f}", (p.get_x() + p.get_width()/2, p.get_height() + 8), ha='center', fontsize=9, fontweight='bold')
axes[0].legend(title='Tipe Hari')
axes[0].spines[['top', 'right']].set_visible(False)

temp_avg = day_clean.groupby('temp_bin', observed=False)['cnt'].mean().reset_index()
colors_temp = ['#d3d3d3', '#6baed6', '#1f77b4']
bars_temp = axes[1].bar(temp_avg['temp_bin'], temp_avg['cnt'], color=colors_temp, width=0.55)
axes[1].set_title('Permintaan Penyewaan Naik Hampir 2x Lipat pada Suhu Warm (>25°C)', fontsize=12, fontweight='bold', pad=12)
axes[1].set_xlabel('Kategori Suhu Harian', fontsize=11)
axes[1].set_ylabel('Rata-rata Penyewaan Harian', fontsize=11)
axes[1].set_ylim(0, 6500)
for bar in bars_temp:
    h = bar.get_height()
    axes[1].annotate(f"{h:,.0f}", (bar.get_x() + bar.get_width()/2, h + 120), ha='center', fontsize=10, fontweight='bold')
axes[1].spines[['top', 'right']].set_visible(False)

plt.tight_layout()
plt.show()
```

## Conclusion & Recommendation

**Conclusion pertanyaan 1:**

Permintaan penyewaan sepeda harian sepanjang periode Januari 2011–Desember 2012 sangat dipengaruhi oleh faktor musim, kondisi cuaca, dan suhu lingkungan:
  - **Berdasarkan Musim (`season`):** Musim **Fall** mencatat rata-rata penyewaan tertinggi sebesar **`5.644` sepeda/hari**, diikuti oleh **Summer** (`4.992` sepeda/hari) dan **Winter** (`4.728` sepeda/hari). Temuan ini selaras dengan hasil *binning* suhu, di mana hari bersuhu **Warm (>25°C)** mencatat rata-rata `5.632` sepeda/hari (**98,78% lebih tinggi** atau hampir 2 kali lipat dibandingkan hari bersuhu **Cool <15°C** sebesar `2.833` sepeda/hari).
  - **Berdasarkan Kondisi Cuaca (`weathersit`):** Kondisi **Clear/Partly Cloudy** menghasilkan rata-rata tertinggi sebesar **`4.877` sepeda/hari**, lalu turun **17,24%** menjadi `4.036` sepeda/hari saat **Mist/Cloudy**, dan anjlok tajam **63,02%** menjadi **`1.803` sepeda/hari** ketika terjadi **Light Snow/Rain**.

**Conclusion pertanyaan 2:**

Pola penyewaan sepeda per jam (`hr`) antara hari kerja dan akhir pekan/hari libur menunjukkan karakteristik waktu dan segmen pengguna yang berlawanan:
  - **Hari Kerja (`workingday = 1`):** Membentuk pola komuter dua puncak (*bimodal*) pada klaster **Morning Rush (06.00–09.00)** dengan rata-rata `278` sepeda/jam (puncak pukul **08.00** sebesar `477` sepeda/jam) dan klaster **Evening Rush (16.00–19.00)** dengan rata-rata `415` sepeda/jam (puncak tertinggi pukul **17.00** sebesar `525` sepeda/jam). Aktivitas ini didominasi oleh pengguna **`registered`** dengan rata-rata `167,6` sepeda/jam (dibandingkan pengguna `casual` hanya `25,6` sepeda/jam).
  - **Akhir Pekan/Hari Libur (`workingday = 0`):** Membentuk pola rekreasi satu puncak landai di siang hari pada klaster **Midday (10.00–15.00)** dengan rata-rata `339` sepeda/jam (puncak pukul **13.00** sebesar `373` sepeda/jam). Pada periode ini, rata-rata pengguna **`casual`** melonjak **2,25 kali lipat (+124,72%)** menjadi `57,4` sepeda/jam dibandingkan hari kerja (`25,6` sepeda/jam).


**Rekomendasi Action Item:**
- **Jadwalkan Pemeliharaan Besar Armada pada Musim Spring dan Hari Hujan:** Pusatkan jadwal perawatan rutin, penggantian komponen, dan inspeksi armada secara intensif selama musim **Spring** serta pada hari dengan prakiraan suhu `<15°C` atau hujan (`Light Snow/Rain`), sehingga 100% armada berada dalam kondisi prima saat menghadapi lonjakan permintaan musim **Summer** dan **Fall**.
- **Terapkan Strategi Promosi Musiman di Musim Spring & Cuaca Mendung:** Luncurkan diskon tarif sewa atau kampanye *bundling* khusus pada musim **Spring** dan hari berawan (**Mist/Cloudy**) untuk menjaga utilisasi armada tetap stabil di luar musim puncak.
- **Lakukan *Rebalancing* Armada Berbasis Klaster Waktu:**
  - Pada **hari kerja**, pastikan distribusi sepeda di area permukiman terisi penuh sebelum pukul **06.00** untuk mengantisipasi *Morning Rush*, lalu lakukan pemindahan stok ke pusat perkantoran dan stasiun transit sebelum pukul **16.00** untuk melayani *Evening Rush*.
  - Pada **akhir pekan/hari libur**, fokuskan penempatan armada di kawasan taman, wisata, dan pusat kota mulai pukul **10.00** hingga **15.00** (*Midday*), disertai penawaran promo konversi keanggotaan (*trial pass*) bagi pengguna **`casual`** agar tertarik beralih menjadi pengguna **`registered`**.
